```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import cm

ef_scale = 1
ef_x = np.linspace(-ef_scale, ef_scale, 101)
ef_y = np.linspace(-ef_scale, ef_scale, 101)

X, Y = np.meshgrid(ef_x, ef_y)


def population(X, Y):
    D = 2
    rho = 1
    population = rho / (2 * np.pi) ** (D / 2) * np.exp(-(X**2 + Y**2) / 2)
    return population


eq_pop = population(X, Y)
# boundary_pop_x = population(np.clip(X, -ef_scale, 0), Y)
boundary_pop_x = population(np.clip(X, 0, ef_scale), Y)

fig, ax = plt.subplots(subplot_kw={"projection": "3d"})
# surf = ax.plot_surface(
#     X, Y, eq_pop, cmap=cm.coolwarm, linewidth=0, alpha=0.5, antialiased=False
# )
surf = ax.plot_surface(
    X,
    Y,
    eq_pop - (boundary_pop_x - eq_pop),
    cmap=cm.coolwarm,
    linewidth=0,
    alpha=0.5,
    antialiased=False,
)
fig.colorbar(surf, shrink=0.5, aspect=5)

plt.show()
```
